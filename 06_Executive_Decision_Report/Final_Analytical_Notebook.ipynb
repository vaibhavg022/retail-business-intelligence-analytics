# Final Analytical Notebook — Retail Business Intelligence Capstone

This notebook consolidates the cleaned retail dataset, KPI results, EDA findings, dashboard-supported metrics, executive insights, and scenario-based recommendations.


```python
import pandas as pd
df = pd.read_csv('../05_Interactive_Dashboard/data/retail-orders-clean.csv')
df['transaction_date'] = pd.to_datetime(df['transaction_date'])
df.head()
```

## Executive KPI Snapshot

- Revenue: ₹1,636,956
- Transactions: 12,575
- Customers: 25
- Average Order Value: ₹130.18
- Quantity: 69,828
- Discount-applied transactions: 33.6%


```python
kpis = {
    'Revenue': df['total_spent'].sum(),
    'Transactions': df['transaction_id'].nunique(),
    'Customers': df['customer_id'].nunique(),
    'AOV': df['total_spent'].sum()/df['transaction_id'].nunique(),
    'Quantity': df['quantity'].sum()
}
kpis
```

```python
category_revenue = df.groupby('category')['total_spent'].sum().sort_values(ascending=False)
year_revenue = df.groupby('year')['total_spent'].sum().sort_index()
discount_revenue = df.groupby('discount_status')['total_spent'].sum()
category_revenue, year_revenue, discount_revenue
```

## Executive Findings

1. Total revenue is ₹1,636,956 across 12,575 transactions.
2. Butchers is the largest revenue category at ₹217,020; Electric Household Essentials is second at ₹214,638.
3. 2024 generated the highest annual revenue in the available dataset at ₹555,192.
4. 33.6% of transactions have a discount applied; discounted transactions contribute ₹553,761 revenue.
5. Cash is the largest payment channel by revenue at ₹567,196.
6. The dataset has no CAC, churn, profit, or geographic fields; those metrics are not estimated or fabricated.


## Strategic Recommendations

**1. Optimize discounting:** test category-level discount thresholds and monitor revenue/AOV before and after promotion.

**2. Prioritize high-revenue segments:** use category and time filters to allocate inventory and promotional attention toward consistently strong segments.

**3. Institutionalize BI monitoring:** maintain the Streamlit dashboard and data-quality checks as a recurring management reporting process.
